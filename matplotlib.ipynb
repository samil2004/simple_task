```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("cleaned_student_data.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

```python
department = df["department"].unique()

sales = df.groupby("department")["marks"].mean()

line_style = dict(
    fontsize=15,
    family="Arial",
    fontweight="bold",
    color="blue"
)

plt.figure(figsize=(8, 5))

plt.bar(department, sales, color="skyblue")

plt.title("Average Student Marks", fontsize=15, color="black")
plt.xlabel("Department", **line_style)
plt.ylabel("Average Marks", **line_style)

plt.show()
```

```python



x = df["department"].value_counts().values

mylabel = df["department"].value_counts().index


plt.pie(
    x,
    labels=mylabel,
    autopct="%1.1f%%"
    
)

plt.title(
    "Students by Department",
    fontsize=20,
    color="red",
    fontweight="bold"
)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["marks"], bins=5, edgecolor="black")

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["attendance"], df["marks"])

plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Attendance vs Marks")

plt.show()
```

```python

import matplotlib.pyplot as plt

x = df["name"]

y = df["marks"]

line_style = dict(
    fontsize=15,
    color="blue",
    fontweight="bold"
)

plt.plot(
    x,
    y,
    color="blue",
    marker="o",
    linestyle="-",
    linewidth=2
)

plt.title(
    "Student Marks",
    fontsize=20,
    color="red",
    fontweight="bold"
)

plt.xlabel("Student Name", **line_style)
plt.ylabel("Marks", **line_style)

plt.xticks(rotation=45)

plt.grid(True)
plt.show()
```
